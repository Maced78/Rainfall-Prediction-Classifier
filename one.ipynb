```python
import pandas as pd
```

```python
d = {'sales': [100000,222000,1000000,522000,111111,222222,1111111,20000,75000,90000,1000000,10000],
'city': ['Tampa','Tampa','Orlando','Jacksonville','Miami','Jacksonville','Miami','Miami','Orlando','Orlando','Orlando','Orlando'],
'size': ['Small', 'Medium','Large','Large','Small','Medium','Large','Small','Medium','Medium','Medium','Small',]}
```

```python
df = pd.DataFrame(data = d)
df['city'].unique()
```

```python
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
one = OneHotEncoder(handle_unknown = 'ignore', sparse_output=False)
```

```python
onehot = one.fit_transform(df[['city']])
columns = one.get_feature_names_out(['city'])
```

```python

onehot_df = (pd.DataFrame(onehot, columns = columns))
onehot_df.head()
```

```python
df = pd.concat([df , onehot_df], axis = 1).drop(columns=['city'])
df
```
